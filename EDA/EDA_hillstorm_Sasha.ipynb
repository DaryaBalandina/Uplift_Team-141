```python
bunch = fetch_hillstrom(target_col='conversion')
dataset = bunch['data']
target = bunch['target']
treatment = bunch['treatment']
dataset['target'] = target
dataset['treatment'] = treatment 
dataset.head()
```

**Описание столбцов**:
1) recency - Количество месяцев с момента последней покупки.
2) history_segment - Категория суммы, потраченной за последний год.
3) history - Фактическая сумма в долларах, потраченная за последний год.
4) mens	- клиент покупал мужские товары за последний год.
5) womens - клиент покупал женские товары за последний год.
6) zip_code	- Классификация почтового индекса: Urban (город), Suburban (пригород) или Rural (сельская местность).
7) newbie - новый клиент за последние двенадцать месяцев.
8) channel - Канал, через который клиент совершал покупки за последний год.
9) target - Конверсия в покупку в течение 2 недель после коммуникации
10) treatment - воздейстаие на пользователя: Womens E-Mail (рассылка с женскими товарами), Mens E-Mail (рассылка с мужскими товарами), No E-Mail (отсутсвие рассылки, то есть КГ)

```python
def build_uplift_ztest_table(df, target_col, control_name='No E-Mail'):

    # 1. Считаем метрики на контроле
    control_df = df[df['treatment'] == control_name]
    n_control = len(control_df)
    conv_control = control_df[target_col].mean()
    success_control = control_df[target_col].sum()
    
    # 2. Находим все тестовые группы
    test_groups = [t for t in df['treatment'].unique() if t != control_name]
    
    results = []
    
    # 3. Считаем Z-тест для каждой тестовой группы против контроля
    for test_name in test_groups:
        test_df = df[df['treatment'] == test_name]
        n_test = len(test_df)
        conv_test = test_df[target_col].mean()
        success_test = test_df[target_col].sum()
        
        # Z-тест (сравниваем count и nobs)
        count = np.array([success_test, success_control])
        nobs = np.array([n_test, n_control])
        stat, pval = proportions_ztest(count, nobs)
        
        results.append({
            'Compare': f"{test_name} vs {control_name}",
            'N_test': n_test,
            'N_control': n_control,
            'Conv_test': conv_test*100,
            'Conv_control': conv_control*100,
            'absolute effect': (conv_test - conv_control)*100,
            'z_stat': stat,
            'p_value_raw': pval
        })
    
    res_df = pd.DataFrame(results)
    
    # 4. Применяем поправку на множественное тестирование
    reject, pvals_corrected, _, _ = multipletests(res_df['p_value_raw'], alpha=0.05, method='holm')
    
    res_df['p_value_holm'] = pvals_corrected
    res_df['is_significant'] = reject
    
    res_df['Conv_test'] = res_df['Conv_test'].round(2).astype(str) + '%'
    res_df['Conv_control'] = res_df['Conv_control'].round(2).astype(str) + '%'
    res_df['absolute effect'] = res_df['absolute effect'].round(2).astype(str) + '%'
    res_df['z_stat'] = res_df['z_stat'].round(3)
    res_df['p_value_raw'] = res_df['p_value_raw'].apply(lambda x: f"{x:.2e}")
    res_df['p_value_holm'] = res_df['p_value_holm'].apply(lambda x: f"{x:.2e}")
    
    return res_df


table_conversion = build_uplift_ztest_table(dataset, target_col='target')
display(table_conversion)
```

###  Баланс числовых фичей

```python
def analyze_numeric_balance(df, treatment_col, numeric_cols, control_name='No E-Mail'):

    # 1. Построение гистограмм
    n_cols = len(numeric_cols)
    n_rows = (n_cols + 1) // 2  # Сетка 2 колонки
    
    fig, axes = plt.subplots(n_rows, 2, figsize=(15, 5 * n_rows))
    axes = axes.flatten()
    
    for i, col in enumerate(numeric_cols):
        ax = axes[i]
        
        sns.histplot(
            data=df, 
            x=col, 
            hue=treatment_col,
            alpha=0.5,      
            ax=ax,
            stat='density',    
            bins=50
        )
        
        ax.set_title(f'Распределение признака: {col}', fontsize=12)
        ax.set_ylabel('Плотность')
        ax.set_xlabel('')
        
    for j in range(i + 1, len(axes)):
        fig.delaxes(axes[j])
        
    plt.tight_layout()
    plt.show()


    # 2. Расчет критериев сбалансированности
    results = []
    control_df = df[df[treatment_col] == control_name]
    test_groups = [t for t in df[treatment_col].unique() if t != control_name]
    
    for group in test_groups:
        test_df = df[df[treatment_col] == group]
        
        for col in numeric_cols:
            x_test = test_df[col].dropna()
            x_control = control_df[col].dropna()
            
            mean_t = x_test.mean()
            mean_c = x_control.mean()
            var_t = x_test.var(ddof=1)
            var_c = x_control.var(ddof=1)
            
            smd = (mean_t - mean_c) / np.sqrt((var_t + var_c) / 2)
            t_stat, p_val = ttest_ind(x_test, x_control, equal_var=False)
            
            results.append({
                'Feature': col,
                'Compare': f"{group} vs {control_name}",
                'Mean_Test': mean_t,
                'Mean_Control': mean_c,
                'SMD': abs(smd),
                'T_Stat': t_stat,
                'P_Value': p_val,
                'Is_Significant': p_val < 0.05
            })
            
    balance_df = pd.DataFrame(results)
    
    # Форматирование таблицы
    balance_df['Mean_Test'] = balance_df['Mean_Test'].round(3)
    balance_df['Mean_Control'] = balance_df['Mean_Control'].round(3)
    balance_df['SMD'] = balance_df['SMD'].round(4)
    balance_df['T_Stat'] = balance_df['T_Stat'].round(3)
    balance_df['P_Value'] = balance_df['P_Value'].apply(lambda x: f"{x:.2e}")
    
    return balance_df.sort_values(by=['Compare', 'Feature'])

# --- ПРИМЕР ИСПОЛЬЗОВАНИЯ ---
numeric_features = ['recency', 'history', 'mens', 'womens', 'newbie']
balance_table = analyze_numeric_balance(dataset, treatment_col='treatment', numeric_cols=numeric_features)
display(balance_table)
```

### Баланс категориальных фичей

```python
def analyze_categorical_balance(df, treatment_col, categorical_cols, control_name='No E-Mail'):

    # 1. Построение столбчатых диаграмм
    n_cols = len(categorical_cols)
    n_rows = (n_cols + 1) // 2
    
    fig, axes = plt.subplots(n_rows, 2, figsize=(15, 5 * n_rows))
    axes = axes.flatten()
    
    for i, col in enumerate(categorical_cols):
        ax = axes[i]
        
        sns.countplot(
            data=df, 
            x=col, 
            hue=treatment_col, 
            stat='percent',
            ax=ax,
        )
        
        ax.set_title(f'Распределение: {col}', fontsize=12)
        ax.set_ylabel('Процент внутри группы (%)')
        ax.set_xlabel('')
        
        # Если категорий много, поворачиваем подписи
        if df[col].nunique() > 4:
            ax.tick_params(axis='x', rotation=45)
            
    for j in range(i + 1, len(axes)):
        fig.delaxes(axes[j])
        
    plt.tight_layout()
    plt.show()

    # 2. Расчет критериев сбалансированности
    results = []
    control_df = df[df[treatment_col] == control_name]
    test_groups = [t for t in df[treatment_col].unique() if t != control_name]
    
    for group in test_groups:
        subset_df = df[df[treatment_col].isin([group, control_name])]
        
        for col in categorical_cols:
            contingency_table = pd.crosstab(subset_df[col], subset_df[treatment_col])
            
            chi2, p_val, _, _ = chi2_contingency(contingency_table)
            
            n = contingency_table.sum().sum()
            min_dim = min(contingency_table.shape) - 1
            cramers_v = np.sqrt(chi2 / (n * min_dim)) if min_dim > 0 else 0
            
            results.append({
                'Feature': col,
                'Compare': f"{group} vs {control_name}",
                'Chi2_Stat': chi2,
                'P_Value': p_val,
                'Cramers_V': abs(cramers_v),
                'Is_Significant': p_val < 0.05
            })
            
    balance_df = pd.DataFrame(results)
    balance_df['Chi2_Stat'] = balance_df['Chi2_Stat'].round(3)
    balance_df['P_Value'] = balance_df['P_Value'].apply(lambda x: f"{x:.2e}")
    balance_df['Cramers_V'] = balance_df['Cramers_V'].round(4)
    return balance_df

categorical_features = ['history_segment', 'zip_code', 'channel']
cat_balance_table = analyze_categorical_balance(dataset, treatment_col='treatment', categorical_cols=categorical_features)
display(cat_balance_table)
```

## Проверка случайности назначения treatment

SMD и хи-квадрат смотрят на признаки по одному и не увидят зависимость, спрятанную во взаимодействиях. Поэтому решается обратная задача: по предтритментным признакам предсказывается принадлежность к тестовой группе. Если назначение было случайным, предсказать его невозможно и ROC AUC должен быть неотличим от 0.5.

```python
import re

import optuna
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import StratifiedKFold, train_test_split

from lightgbm import LGBMClassifier
BOOSTER = 'lightgbm'

RANDOM_STATE = 42
N_TRIALS = 50
INNER_SPLITS = 3
EVAL_SPLITS = 5
TUNE_SIZE = 0.3
N_BOOT_AUC = 1000
ALPHA = 0.05

N_CPU = 10
N_JOBS_OPTUNA = max(1, min(8, N_CPU // 2))
N_JOBS_MODEL = max(1, N_CPU // N_JOBS_OPTUNA)

optuna.logging.set_verbosity(optuna.logging.WARNING)

CONTROL_NAME = 'No E-Mail'
TEST_GROUPS = ['Mens E-Mail', 'Womens E-Mail']

PRETREATMENT_FEATURES = [
    'recency', 'history', 'mens', 'womens', 'newbie',
    'history_segment', 'zip_code', 'channel',
]


def make_xy(df, test_name, control_name=CONTROL_NAME):
    sub = df[df['treatment'].isin([test_name, control_name])]
    X = pd.get_dummies(sub[PRETREATMENT_FEATURES], drop_first=True, dtype=float).reset_index(drop=True)
    # Категории history_segment содержат запятые ('7) $1,000 +'), а LightGBM не
    # принимает служебные JSON-символы в именах фичей — чистим имена колонок.
    X.columns = [re.sub(r'[^0-9A-Za-z_]+', '_', str(c)).strip('_') for c in X.columns]
    y = (sub['treatment'] == test_name).astype(int).to_numpy()
    return X, y


def make_model(params, n_jobs=N_JOBS_MODEL):
    if BOOSTER == 'lightgbm':
        return LGBMClassifier(
            random_state=RANDOM_STATE,
            n_jobs=n_jobs,
            verbose=-1,
            subsample_freq=1,
            **params,
        )


def suggest_params(trial):
    return dict(
        n_estimators=trial.suggest_int('n_estimators', 100, 400, step=50),
        learning_rate=trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
        num_leaves=trial.suggest_int('num_leaves', 7, 63),
        max_depth=trial.suggest_int('max_depth', 2, 8),
        min_child_samples=trial.suggest_int('min_child_samples', 20, 300),
        subsample=trial.suggest_float('subsample', 0.6, 1.0),
        colsample_bytree=trial.suggest_float('colsample_bytree', 0.6, 1.0),
        reg_lambda=trial.suggest_float('reg_lambda', 1e-3, 10.0, log=True),
    )


def cv_auc(X, y, params, n_splits=INNER_SPLITS, seed=RANDOM_STATE, n_jobs_model=N_JOBS_MODEL):
    skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=seed)
    scores = []
    for tr, va in skf.split(X, y):
        model = make_model(params, n_jobs=n_jobs_model)
        model.fit(X.iloc[tr], y[tr])
        scores.append(roc_auc_score(y[va], model.predict_proba(X.iloc[va])[:, 1]))
    return float(np.mean(scores))
```

```python
def tune_params(X, y, n_trials=N_TRIALS, inner_splits=INNER_SPLITS, seed=RANDOM_STATE):
    """Подбор гиперпараметров Optuna на тюнинговой части выборки.
    Параллелизм — по trials."""
    study = optuna.create_study(
        direction='maximize',
        sampler=optuna.samplers.TPESampler(seed=seed),
    )
    study.optimize(
        lambda t: cv_auc(X, y, suggest_params(t), n_splits=inner_splits, seed=seed),
        n_trials=n_trials,
        n_jobs=N_JOBS_OPTUNA,
        show_progress_bar=False,
    )
    return study.best_params, float(study.best_value)


randomization = {}

for test_group in TEST_GROUPS:
    X, y = make_xy(dataset, test_group)

    # Тюнинговая и оценочная части не пересекаются: гиперпараметры выбираются по
    # настоящим меткам, поэтому оценивать AUC на тех же данных нельзя.
    X_tune, X_eval, y_tune, y_eval = train_test_split(
        X, y, train_size=TUNE_SIZE, stratify=y, random_state=RANDOM_STATE,
    )
    X_eval = X_eval.reset_index(drop=True)

    best_params, best_inner_auc = tune_params(X_tune.reset_index(drop=True), y_tune)

    randomization[test_group] = {
        'X_eval': X_eval,
        'y_eval': y_eval,
        'params': best_params,
        'tune_auc': best_inner_auc,
    }
    print(f'{test_group} vs {CONTROL_NAME}: n_tune={len(y_tune)}, n_eval={len(y_eval)}, '
          f'лучший AUC на тюнинговой части={best_inner_auc:.4f}')
    print('  params:', best_params)
```

```python
def oof_auc_fixed(X, y, params, n_splits=EVAL_SPLITS, seed=RANDOM_STATE, n_jobs_model=N_JOBS_MODEL):
    """OOF AUC на оценочной части с фиксированными гиперпараметрами."""
    skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=seed)
    oof = np.full(len(y), np.nan)
    for tr, te in skf.split(X, y):
        model = make_model(params, n_jobs=n_jobs_model)
        model.fit(X.iloc[tr], y[tr])
        oof[te] = model.predict_proba(X.iloc[te])[:, 1]
    return float(roc_auc_score(y, oof)), oof


def bootstrap_auc_ci(y, scores, n_boot=N_BOOT_AUC, alpha=ALPHA, seed=RANDOM_STATE):
    """Перцентильный бутстрап AUC. Ресэмпл стратифицирован по классу, чтобы в
    каждой итерации сохранялся размер обеих групп. Интервал учитывает разброс
    самой метрики на этой выборке, но не разброс обучения модели."""
    rng = np.random.default_rng(seed)
    idx_pos = np.flatnonzero(y == 1)
    idx_neg = np.flatnonzero(y == 0)
    aucs = np.empty(n_boot)
    for b in range(n_boot):
        idx = np.concatenate([
            rng.choice(idx_pos, idx_pos.size, replace=True),
            rng.choice(idx_neg, idx_neg.size, replace=True),
        ])
        aucs[b] = roc_auc_score(y[idx], scores[idx])
    lo, hi = np.quantile(aucs, [alpha / 2, 1 - alpha / 2])
    return float(lo), float(hi), aucs


rows = []

for test_group, res in randomization.items():
    auc_obs, oof = oof_auc_fixed(res['X_eval'], res['y_eval'], res['params'])
    lo, hi, boot_aucs = bootstrap_auc_ci(res['y_eval'], oof)

    res.update({'auc': auc_obs, 'oof': oof, 'ci': (lo, hi), 'boot_aucs': boot_aucs})

    rows.append({
        'Compare': f'{test_group} vs {CONTROL_NAME}',
        'N_eval': len(res['y_eval']),
        'AUC': round(auc_obs, 4),
        'CI_95_low': round(lo, 4),
        'CI_95_high': round(hi, 4),
        'CI_covers_0.5': bool(lo <= 0.5 <= hi),
    })

randomization_table = pd.DataFrame(rows)
display(randomization_table)

fig, axes = plt.subplots(1, len(randomization), figsize=(7 * len(randomization), 4.5), squeeze=False)
for ax, (test_group, res) in zip(axes[0], randomization.items()):
    ax.hist(res['boot_aucs'], bins=30, color='lightgrey', edgecolor='black',
            label='Бутстрап-распределение AUC')
    ax.axvline(res['auc'], color='red', linestyle='--', linewidth=2,
               label=f"AUC = {res['auc']:.4f}")
    ax.axvline(0.5, color='blue', linestyle=':', linewidth=1.5, label='AUC = 0.5')
    ax.set_title(f'{test_group} vs {CONTROL_NAME}\n'
                 f'95% ДИ: [{res["ci"][0]:.4f}; {res["ci"][1]:.4f}]')
    ax.set_xlabel('OOF ROC AUC на оценочной части')
    ax.set_ylabel('Частота')
    ax.legend(fontsize=8)
plt.suptitle('Проверка случайности назначения treatment', fontsize=15, fontweight='bold')
plt.tight_layout()
plt.show()
```

ROC AUC моделей, предсказывающих тритмент примерно равен 0.5 - делаем вывод о том, что тритмент назначался случайно.

```python
df_plot = dataset.copy()

features_for_bucket = ['recency', 'history']
features_no_bucket = ['history_segment', 'mens', 'womens', 'zip_code', 'newbie', 'channel']
all_features = features_for_bucket + features_no_bucket

control_name = 'No E-Mail'
test_groups = ['Mens E-Mail', 'Womens E-Mail']

N_BOOT_CI = 2000
ALPHA_CI = 0.05
rng_ci = np.random.default_rng(42)


def uplift_bootstrap_ci(stats, group, control_name, n_boot=N_BOOT_CI, alpha=ALPHA_CI, rng=rng_ci):
    """Перцентильный бутстрап-ДИ разности конверсий по каждому бину.
    Для 0/1 таргета непараметрический бутстрап среднего в точности эквивалентен
    Binomial(n, p_hat) / n, поэтому матрицу ресэмплов хранить не нужно.
    stats — результат agg(['sum', 'count']).unstack() по (бин, treatment)."""
    out = {}
    for name in (group, control_name):
        s = stats[('sum', name)].to_numpy(dtype=float)
        n = stats[('count', name)].to_numpy(dtype=float)
        ok = n > 0
        p = np.divide(s, n, out=np.zeros_like(s), where=ok)
        draws = rng.binomial(np.where(ok, n, 1).astype(int), p, size=(n_boot, n.size))
        out[name] = np.divide(draws, n, out=np.full_like(draws, np.nan, dtype=float), where=ok)

    boot = out[group] - out[control_name]
    low, high = np.quantile(boot, [alpha / 2, 1 - alpha / 2], axis=0)
    return low, high


for feature in all_features:

    if feature in features_for_bucket:
        col_name = f'bucket_{feature}'
        df_plot[col_name] = pd.qcut(df_plot[feature], q=10, duplicates='drop')
        x_label = f'Бакеты {feature}'
    else:
        col_name = feature
        x_label = f'Значения {feature}'

    pivot = df_plot.groupby([col_name, 'treatment'], observed=False)['target'].mean().unstack()
    stats = df_plot.groupby([col_name, 'treatment'], observed=False)['target'].agg(['sum', 'count']).unstack()

    fig, axes = plt.subplots(1, 2, figsize=(16, 6), sharey=True)
    x_axis = range(len(pivot))

    for i, group in enumerate(test_groups):
        ax = axes[i]

        uplift = pivot[group] - pivot[control_name]
        ci_low, ci_high = uplift_bootstrap_ci(stats, group, control_name)
        yerr = np.vstack([uplift.to_numpy() - ci_low, ci_high - uplift.to_numpy()])
        yerr = np.nan_to_num(np.clip(yerr, 0, None))

        bars = ax.bar(x_axis, uplift, color='skyblue' if i == 0 else 'lightcoral', edgecolor='black', zorder=2)
        ax.errorbar(x_axis, uplift, yerr=yerr, fmt='none', ecolor='black',
                    elinewidth=1.3, capsize=4, zorder=3)

        ax.axhline(0, color='red', linestyle='--', linewidth=1, zorder=1)
        ax.set_xticks(x_axis)
        ax.set_xticklabels([str(idx) for idx in pivot.index], rotation=45, ha='right')
        ax.set_xlabel(x_label)

        if i == 0:
            ax.set_ylabel('Uplift (Разница конверсий)')

        ax.set_title(f'Uplift по {feature} ({group} vs {control_name})')

        for bar, lo, hi in zip(bars, ci_low, ci_high):
            yval = bar.get_height()
            if not np.isfinite(yval):
                continue
            va = 'bottom' if yval >= 0 else 'top'
            offset = (hi - yval) if yval >= 0 else (lo - yval)
            offset = offset if np.isfinite(offset) else 0.0
            ax.text(bar.get_x() + bar.get_width() / 2, yval + offset, f'{yval:.4f}',
                    ha='center', va=va, fontsize=8)

    plt.suptitle(f'Гетерогенность эффекта: {feature} (усы — 95% бутстрап-ДИ uplift)',
                 fontsize=16, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.show()
```

**Выводы по анализу**

1) Тест проведён корректно. Разницы по отдельным признакам между группами нет: SMD < 0.1, хи-квадрат незначим. Дополнительно это проверено бустингом, предсказывающим treatment по предтритментным признакам: OOF AUC 0.4991 с 95% бутстрап-ДИ [0.4926; 0.5055] для мужской рассылки и 0.4968 с ДИ [0.4900; 0.5029] для женской — оба интервала накрывают 0.5, то есть группы по признакам неразличимы. Рандомизацию не отвергаем, эффект можно оценивать сырой разностью конверсий, без поправки на признаки.

2) Эффект от коммуникаций есть. Получаем стат значимой прирост в конверсию как для коммуникаций с женскими товарами, так и с мужскими. При чем от мужский коммуникаций эффект значительно выше

3) Сильных линейных взаимосвязей с таргетом нет. Поэтому в моделях T-learner, S-learner и т.п. лучше использовать нелинейные модели (бустинги, случайные деревья). Также можно удалить категориальный признак history_segment, так как он зависит от признака history

4) При расчете uplift по признакам можно увидеть, что люди с более давними последними попупками (recency) более склонны к покупке после воздействия. Для мужских коммуникаций на всех бинах uplift положительный, для женских есть отрицательный (коммуникации негативно влияют на этот сегмент). Также можно заметить, что рассылка с женскими товарами сильнее влияет на тех людей, кто имел покупки этих товаров в прошлом (для мужских товаров это тоже свойственно, но uplift слабее), а также на новичков. Содержательно опираться стоит только на бины, где 95% бутстрап-ДИ (усы на графиках) не накрывает ноль: при базовой конверсии около процента бин из нескольких сотен наблюдений даёт очень широкий интервал, и знак столбика сам по себе ничего не доказывает.

5) Фичи по разному влияют на покупку в зависимости от воздействия, поэтому в качестве бейзлайна я бы оставил все, кроме history_segment
